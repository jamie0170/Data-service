```python
df1 = (spark.read
    .option("header", "true")
    .csv("/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv"))

# Delta disallows spaces/special chars in column names — sanitize before writing
df1 = df1.toDF(*[c.strip().replace(" ", "_").replace("(", "").replace(")", "") for c in df1.columns])
display(df1.limit(5))

df1.write.mode("overwrite").saveAsTable("default.mobile_os_usage")
```
